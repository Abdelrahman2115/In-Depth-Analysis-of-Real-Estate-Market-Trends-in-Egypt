```python
null_array = round(df.isnull().sum()/len(df)*100,2).sort_values().to_frame().rename(columns= {0:"Train % of Missing Values"})
print(null_array, end='\n')
```

```python
# Fill baths and beds with median (more robust than mean)
df = df.dropna(subset=['baths'])
df = df.dropna(subset=['beds'])
```

```python
# For down_payment: fill missing with 0 (assuming no down payment if missing)
df['down_payment'] = df['down_payment'].fillna(0)
```

```python
# For plan_type: fill missing with 'unknown' (new category)
df['plan_type'] = df['plan_type'].fillna('unknown')
```

```python
# Optionally, create a binary flag for down_payment missingness if useful
df['down_payment_missing'] = (df['down_payment'] == 0).astype(int)
```

```python

print(df.isnull().mean() * 100)  # Confirm missing data percentages now
```

```python
df.to_csv('real_estate_data_bayut_full_cleaned_data.csv', index=False)
```
